```python
import pandas as pd
import numpy as np
```

```python
reviews_df = pd.read_csv("reviews_sample.csv")
reviews_df.head()
```

```python
import re

STOPWORDS = {"and", "a", "the"}


def preprocess_text(text, remove_stopwords=True):
    """Lowercase, remove punctuation, tokenize, and optionally remove stopwords."""
    text = "" if pd.isna(text) else str(text)
    text = text.lower()
    text = re.sub(r"[^\w\s]", " ", text)
    tokens = text.split()

    if remove_stopwords:
        tokens = [token for token in tokens if token not in STOPWORDS]

    return tokens


reviews_df["text"] = reviews_df["review_body"].fillna(reviews_df["title"])
reviews_df["tokens"] = reviews_df["text"].apply(preprocess_text)
reviews_df["processed_text"] = reviews_df["tokens"].str.join(" ")

reviews_df[["text", "tokens", "processed_text"]].head()
```
