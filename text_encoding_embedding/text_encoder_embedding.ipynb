```python
from sklearn.preprocessing import OneHotEncoder
import numpy as np
```

```python
documents=['people watch movie,'
'people watch cricket',
'people like movie',
'people like cricket']
```

```python
token=[i.lower().split() for i in documents]
```

```python
token
```

```python
all_words=[[word] for sentence in token for word in sentence]
```

```python
all_words
```

```python
encode=OneHotEncoder(sparse_output=False)
```

```python
x=encode.fit_transform(all_words)
```

```python
x
```

```python
import pandas as pd
```

```python
data=pd.DataFrame(x,columns=encode.get_feature_names_out(['all_words']))
```
